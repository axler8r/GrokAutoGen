```python
import os
from typing import Iterator
from dotenv import load_dotenv
from secretstorage import (
    dbus_init,
    get_default_collection,
    DBusConnection,
    Collection,
    Item,
)


connection: DBusConnection = dbus_init()
collection: Collection = get_default_collection(connection)
collection.unlock()
autogen_secrets: Iterator[Item] = collection.search_items(
    {"provider": "OpenAI", "project": "GrokAutoGen"}
)

openai_api_key = None
if autogen_secrets:
    for secret in autogen_secrets:
        openai_api_key: bytes = secret.get_secret()
    connection.close()
elif load_dotenv():
    openai_api_key = os.getenv("OPENAI_API_KEY").encode()
else:
    print("No OpenAI API key found in environment variables or secret storage.")

print(f"OpeanAI API Key: {openai_api_key.decode().strip()}")
```
